```python
import requests
from bs4 import BeautifulSoup
import pandas as pd


# Obtener el contenido HTML de la página
response = requests.get("https://evtol.news/aircraft")
soup = BeautifulSoup(response.content, "html.parser")

# Buscar los links
links = [a["href"] for a in soup.find_all("a", href=True) if "https://evtol.news" in a["href"] and not "concept" in a["href"] and not "demonstrator" in a["href"] and not "defunct" in a["href"]]
count = 0
# Imprimir los links
for link in links:
    #print(link)
    count+=1
print(links)
print(count)
```

```python

# Definir una lista para almacenar datos extraídos
extracted_data = []

for link in links:
    # Obtener el contenido HTML de cada enlace
    response = requests.get(link)
    soup = BeautifulSoup(response.content, "html.parser")

    # Encontrar la lista de especificaciones
    specs_list = soup.find_all("li")

    capacity = None
    max_range = None

for item in specs_list:
    # Extract capacity
    if "Capacity" in item.text:
        capacity_parts = item.text.split(":")
        if len(capacity_parts) >= 2:
            capacity_text = capacity_parts[1].strip()
            capacity = capacity_text

    # Extract maximum range
    elif "Maximum range" in item.text:
        max_range_parts = item.text.split(":")
        if len(max_range_parts) >= 2:
            max_range = max_range_parts[1].strip()

    # Handle missing values (optional)
    if capacity is None:
        capacity = "Not Available"
    if max_range is None:
        max_range = "Not Available"

    # Crear un diccionario para datos extraídos
    data = {"link": link, "capacity": capacity, "max_range": max_range}

    # Agregar datos a la lista
    extracted_data.append(data)

# Crear un DataFrame de pandas
df = pd.DataFrame(extracted_data)

# Imprimir el DataFrame
print(df)
```
